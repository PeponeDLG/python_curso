7. Frecuencia de palabras
texto = "hoy es un gran día porque hoy empezamos un nuevo proyecto en Python"
Cuenta cuántas veces aparece cada palabra y descarta con un comprehension las que tengan
menos de 3 letras.

```python
texto = "hoy es un gran día porque hoy empezamos un nuevo proyecto en Python"

palabras = texto.split()
frecuencia = {}

for palabra in palabras:
    if palabra in frecuencia:
        frecuencia[palabra] += 1
    else:
        frecuencia[palabra] = 1

frecuencia_filtrada = {palabra: frecuencia for palabra, frecuencia in frecuencia.items() if len(palabra) >= 3}

print(frecuencia_filtrada)
```

8. Ranking de ventas
ventas = {
"Ana": 1200,
"Luis": 980,
"Marta": 1420,
"Jorge": 750,
"Lucía": 1320
}
Ordena de mayor a menor usando lambda y muestra los 3 primeros.

```python
ventas = {
    "Ana": 1200,
    "Luis": 980,
    "Marta": 1420,
    "Jorge": 750,
    "Lucía": 1320
}

ventas_ordenadas = sorted(ventas.items(), key=lambda item: item[1], reverse=True)

top_3 = ventas_ordenadas[:3]

for i, (nombre, venta) in enumerate(top_3, 1):
    print(f"{i}. {nombre}: {venta}")
```

9. Posiciones de cada elemento

14. Facturación total
precios = {"barra": 1.10, "leche": 0.95, "huevos": 2.20}
ventas = {"barra": 30, "leche": 12, "huevos": 6}
Calcula la facturación total (precio × unidades).

```python
precios = {"barra": 1.10, "leche": 0.95, "huevos": 2.20}
ventas = {"barra": 30, "leche": 12, "huevos": 6}

facturacion_total = sum(precios[producto] * ventas[producto] for producto in precios.keys() & ventas.keys())

print(f"Facturación total: {facturacion_total} €")
```

18. Ordenar tuplas
personas = [
("Ana", 29, 167),
("Luis", 22, 173),
("Marta", 29, 165),
("Jorge", 22, 182),
("Lucía", 29, 167)
]
Ordena por:
edad descendente
altura descendente
nombre ascendente

```python
personas = [("Ana", 29, 167), ("Luis", 22, 173), ("Marta", 29, 165), 
            ("Jorge", 22, 182), ("Lucía", 29, 167)]

personas_ordenadas = sorted(personas, key=lambda x: (-x[1], -x[2], x[0]))

for nombre, edad, altura in personas_ordenadas:
    print(f"{nombre:6} {edad:2} años {altura:3} cm")
```

19. Filtrado doble con comprensión
numeros = [3, 8, 11, 14, 4, 2, 9, 21, 10]
Quédate solo con los pares cuyo cuadrado sea mayor que 50.

```python
numeros = [3, 8, 11, 14, 4, 2, 9, 21, 10]

resultado = [n for n in numeros if n % 2 == 0 and n ** 2 > 50]

print(resultado)
```
