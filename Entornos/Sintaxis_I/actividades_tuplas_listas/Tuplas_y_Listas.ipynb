**7. Clasificación por tramos (no IRPF)**
Tupla:
tramos = (
(20, "bajo"),
(50, "medio"),
(80, "alto")
)
Valor: 47
Tareas:
• Determina a qué categoría pertenece el valor.
• Explica el resultado.

```python
tramos = (
(20, "bajo"),
(50, "medio"),
(80, "alto")
)
valor = 51
for i in reversed(tramos):
    if valor >= i[0]:
        print(i[1])
        break
## Cada tupla indica el valor mínimo que encuadra al valor dado dentro de la categoría. Por lo que si hacemos el bucle a la inversa, nunca podrá estar dentro de la 
## categoría superior. Y al escapar del bulce al encontrar su categoría, tampoco pasará a una categoría inferior.
```

**8. Creación de códigos de producto**

```python
letras = ("K", "L", "M")
digitos = (0, 1, 2, 3)
lista1=[]

for i in letras:
    for j in digitos:
        lista1.append(i+str(j))

print(lista1)
len(lista1)
```

**12. Ordenación múltiple de nombres**
Lista:
nombres = ["Ana", "Carmen", "Luis", "Bea"]
Tareas:
• Ordénalos por longitud.
• Ordénalos alfabéticamente.
• Ordénalos por su última letra.

```python
nombres = ["Ana", "Carmen", "Luis", "Bea"]
nombres.sort(key=len)
print(nombres)
nombres.sort(key=str.capitalize)
print(nombres)
nombres.sort(key=lambda x: x[-1])
print(nombres)
```

**14. Cálculo de precios con descuento**
Lista:
precios = [50, 80, 120, 200]
Tareas:
• Calcula una nueva lista con un 10% de descuento usando un bucle.
• Haz lo mismo en una sola línea con list comprehension

```python
precios = [50, 80, 120, 200]

for i,j in enumerate(precios):
    print(int(j-(j*0.10)))

print([int(p - (p * 0.10)) for p in precios])
```

**15. Filtrado de temperaturas peligrosas**

• Crea lista de temperaturas superiores a 30.
• Únelas en una lista final.
