```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Ejemplo de datos
datos = pd.DataFrame({"Experiencia": [1, 2, 3, 4, 5]})

# ANTES de reshape
x_1d = datos["Experiencia"].values
print("ANTES de reshape:")
print(f"  Valores: {x_1d}")
print(f"  Forma: {x_1d.shape}")  # (5,)
print(f"  Dimensiones: {x_1d.ndim}D")

# DESPUÉS de reshape
x_2d = datos["Experiencia"].values.reshape((-1, 1))
print("\nDESPUÉS de reshape:")
print(f"  Valores:\n{x_2d}")
print(f"  Forma: {x_2d.shape}")  # (5, 1)
print(f"  Dimensiones: {x_2d.ndim}D")

# Crear figura
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Plot 1D
axes[0].plot(x_1d, 'o-')
axes[0].set_title(f'1D: {x_1d.shape}')
axes[0].set_xlabel('Índice')
axes[0].set_ylabel('Valor')

# Plot 2D
axes[1].plot(x_2d, 'o-')
axes[1].set_title(f'2D: {x_2d.shape}')
axes[1].set_xlabel('Índice')
axes[1].set_ylabel('Valor')
axes[1].legend(['Columna 0'])

plt.tight_layout()
plt.show()
```
