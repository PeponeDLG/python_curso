## 7. Clasificación por tramos (no IRPF)

Tupla:
```py

(20, "bajo"),
(50, "medio"),
(80, "alto")
)

Valor: 47
```

Tareas:

- Determina a qué categoría pertenece el valor.

- Explica el resultado.

````python
tramos = (
(20, "bajo"),
(50, "medio"),
(80, "alto")
)

valor = 47

# Determina a qué categoría pertenece el valor
if valor <= tramos[0][0]:
    print(f"Categoría: bajo.")
elif valor <= tramos[1][0]:
    print(f"Categoría: medio.")
elif valor <= tramos[2][0]:
    print(f"Categoría: alto.")
else:
    print(f"Categoría: Fuera de rango.")

# Según la tupla, si el rango es 20 es bajo, por tanto considero que por debajo
# de 20 sigue siendo bajo, entre 21 y 50 medio y entre 51 y 80 alto.
````

## 12. Ordenación múltiple de nombres

Lista:
```py

```

Tareas:

- Ordénalos por longitud.

- Ordénalos alfabéticamente.

- Ordénalos por su última letra.

````python
nombres = ["Ana", "Carmen", "Luis", "Bea"]

# Ordena por longitud
print(f"Nombres ordenados por longitud: {sorted(nombres, key=len)}")

# Ordena alfabéticamente
print(f"Nombres ordenados alfabéticamente: {sorted(nombres)}")

# Ordena por su última letra
print(f"Nombres ordenados por su última letra: {sorted(nombres, key=lambda x:x[-1])}")
````

## 14. Cálculo de precios con descuento

Lista:
```py
precios = [50, 80, 120, 200]
```

Tareas:

- Calcula una nueva lista con un 10% de descuento usando un bucle.

- Haz lo mismo en una sola línea con list comprehension.

````python
precios = [50, 80, 120, 200]

# Calcula una nueva lista con un 10% de descuento usando un bucle
descuento_1 = []

for elem in precios:
    descuento_1.append(elem * 0.9)

print(f"Lista con descuento (bucle): {descuento_1}")

# Con list comprehension
descuento_2 = [elem * 0.9 for elem in precios]

print(f"Lista con descuento (list comprehension): {descuento_2}")
````

## 15. Filtrado de temperaturas peligrosas

Lista:
```py
lecturas = [-3, 5, 12, 40, -1, 33]
```

Tareas:

- Crea lista de temperaturas bajo cero.

- Crea lista de temperaturas superiores a 30.

- Únelas en una lista final.

````python
lecturas = [-3, 5, 12, 40, -1, 33]

# Crea lista de temperaturas bajo cero
bajo_cero = [temp for temp in lecturas if temp < 0]

print(f"Lista de temperaturas bajo 0: {bajo_cero}")

# Crea lista de temperaturas superiores a 30
sobre_30 = [temp for temp in lecturas if temp > 30]

print(f"Lista de temperaturas superiores a 30: {sobre_30}")

# Lista final uniendo las 3
lista_final = lecturas + bajo_cero + sobre_30
print(lista_final)
````

# 18. Eliminación de duplicados sin sets

Lista:
```py
datos = [3, 3, 2, 5, 2, 2, 8]
```

Tareas:

- Crea una lista sin duplicados manteniendo el orden.

- No usar set().

````python
datos = [3, 3, 2, 5, 2, 2, 8]

# Crea una lista sin duplicados manteniendo el orden
sin_duplicados = []
for elem in datos:
    if elem not in sin_duplicados:
        sin_duplicados.append(elem)

print(f"Lista sin duplicados: {sin_duplicados}")
````

## 19. Simulación de pila y cola en un turno

Lista vacía:
```py
turno = []
```

Tareas:

- Encola los valores: 101, 102, 103.

- Desencola el primero.

- En la misma lista, simula una pila añadiendo: 501, 502, 503.

- Extrae el último elemento.

````python
turno = []

# Encola los valores: 101, 102, 103
cola = [101, 102, 103]
# turno = [turno.append(elem) for elem in cola]     # Investigar por qué añade 3 None en vez de los elementos

for elem in cola:
    turno.append(elem)

print(turno)

# Desencola el primero
turno.pop(0)
print(turno)

# En la misma lista, simula una pila añadiendo: 501, 502, 503
pila = [501, 502, 503]
for elem in pila:
    turno.append(elem)

print(turno)

# Extrae el último elemento
turno.pop()
print(turno)
````
