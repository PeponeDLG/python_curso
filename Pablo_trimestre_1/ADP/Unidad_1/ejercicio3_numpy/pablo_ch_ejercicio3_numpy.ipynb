## A partir del siguiente código:

```python
import numpy as np

nombres = np.array([
    "Alberto",
    "Álvaro",
    "Antonio",
    "Guillermo",
    "José Manuel",
    "Juan Diego",
    "Juan Ignacio",
    "Juan José",
    "María",
    "Raúl",
    "Victor"
])

print(nombres)
notas = np.random.randint(1, 11, nombres.size)
print(notas)
```

## Calcula:
1. Nota media de los alumnos

```python
# Calcula la media usando np.mean()
print("-"*100)
print(f"Nota media de los alumnos: {np.mean(notas)}")
```

2. Desviación estándar

```python
# Calcula la desviación estándar usando np.std()
print(f"Desviación estándar: {np.std(notas)}")

# Por ejemplo una desviación estándar de 2 significa que la mayoría de las notas están
# distribuidas en un rango de 2 puntos arriba y abajo de la media. No sería un dato
# representativo (por encima de 0.2 no es representativa), es decir la muestra está muy
# dispersa.
```

3. Número de aprobados

```python
# Versión larga para entender el funcionamiento de where:

# Comprueba si las notas son iguales o superiores a 5 (establece una condición)
# condicion = notas >= 5

# Filtra según la condición
# result = np.where(condicion) # El primer elemento de result es una tupla con los datos filtrados

# Cuenta los elementos filtrados para mostrar el resultado
# print(f"Número de aprobados: {result[0].size}")


# Versión refactorizada usando where():

# Cuenta los elementos que pasan la condición de ser mayores o iguales a 5 y muestra el número
# print(f"Número de aprobados: {np.where(notas >= 5)[0].size}")


# Versión usando indexación booleana:
print(nombres[notas >= 5].size)
```

4. Mostrar nombres de aprobados

```python
# Versión larga para entender el funcionamiento de where:

# Establece la condición sobre el array de notas
# El primer elemento que devuelve where() es una tupla de índices que han pasado la condición
# aprobados = np.where(notas >= 5)

# Utiliza la tupla de índices de notas que han pasado la condición para mostrar los nombres de
# los aprobados seleccionando con dichos índices los elementos en el array de nombres
# print(nombres[aprobados[0]])


# Versión refactorizada:

# Muestra los nombres que corresponden a las notas que pasan la condición de ser mayores o iguales a 5
# print(f"Nombres de aprobados: {nombres[np.where(notas >= 5)]}")


# Versión usando indexación booleana:
print(nombres[notas >= 5])
```

5. Mostrar nombres con nota superior a la media

```python
# Muestra los nombres que corresponden a las notas que pasan la condición de ser superiores a la media
# print(f"Nombres de alumnos con nota superior a la media: {nombres[np.where(notas > np.mean(notas))]}")

# Versión usando indexación booleana:
print(f"Alumnos con nota superior a la media: {nombres[notas > np.mean(notas)]}")
```

6. Mostrar nombres del revés mediante una slice

```python
# Invierte el array usando slice sobre el array entero, elemento a elemento y aplicando orden inverso
print(f"Alumnos por orden alfabético descendente: {nombres[::-1]}")
```

7. Ordenar, haciendo una copia de datos, las notas de menor a mayor

```python
copia_notas = notas.copy()  # Crea una copia del array notas
print(f"Notas: {copia_notas}")
print(f"Notas ordenadas de menor a mayor: {np.sort(copia_notas)}")
```

8. Asignar un 5 a los que tienen un 4

```python
print(f"Notas: {notas}")
print(f"Notas con 4 subidas a 5: {np.where(notas == 4, 5, notas)}")
```

9. Sumar 1 a todas las notas

```python
print(f"Notas: {notas}")
print(f"Notas + 1: {np.add(notas, 1)}")
```

10. Sumar 1 a todas las notas mayores a 4

```python
print(f"Notas: {notas}")
print(f"Notas mayores a 4 subidas 1 punto: {np.where(notas > 4, np.add(notas, 1), notas)}")
```
