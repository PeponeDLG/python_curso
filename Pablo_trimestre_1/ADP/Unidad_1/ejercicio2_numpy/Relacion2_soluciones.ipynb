1. Crear una función “inicializar_array” que reciba por parámetro un array y devuelva el
array inicializado con valores introducidos por teclado.

```python
import numpy as np

# Función que inicializa un array con valores introducidos por teclado
def inicializar_array(arr):
    for i in range(arr.size):
        valor = int(input(f"Introduce valor entero para la posición {i}: "))
        arr[i] = valor
    return arr

# Preguntamos al usuario el tamaño del array
tam = int(input("¿Cuántos elementos tendrá el array? "))

# Creamos el array con ceros y tipo entero
mi_array = np.zeros(tam, dtype=int)

# Inicializamos el array con valores introducidos por el usuario
mi_array = inicializar_array(mi_array)

# Mostramos el array final
print("Array inicializado:", mi_array)
```

2. Crear las funciones “media”, “máximo”, “mínimo” que reciba un array de valores
numéricos y devuelva la media aritmética, el valor máximo y el valor mínimo
respectivamente. Realizar el cálculo manualmente recorriendo el array con un for.

```python
# Función para calcular la media aritmética
def media(arr):
    suma = 0
    # Sumamos todos los elementos del array
    for valor in arr:
        suma += valor
    return suma / arr.size  # Dividimos entre el número de elementos

# Función para calcular el valor máximo
def máximo(arr):
    max_val = arr[0]  # Inicializamos con el primer valor
    for valor in arr:
        if valor > max_val:
            max_val = valor  # Actualizamos si encontramos uno mayor
    return max_val

# Función para calcular el valor mínimo
def mínimo(arr):
    min_val = arr[0]  # Inicializamos con el primer valor
    for valor in arr:
        if valor < min_val:
            min_val = valor  # Actualizamos si encontramos uno menor
    return min_val

# Ejemplo de uso
datos = np.array([4, 7, 2, 9, 5])
print("Array:", datos)
print("Media:", media(datos))
print("Máximo:", máximo(datos))
print("Mínimo:", mínimo(datos))
```

3. Crear la función “encontrar” (usando for) que reciba dos parámetros: un array
numérico y un valor. La función debe buscar en el array si el “valor” se encuentra en
él. La función debe devolver:
○ Si valor se encuentra en el array, devolver el índice (primera aparición)
○ Si el valor no se encuentra en el array, devolver -1.

```python
# Función que busca un valor en el array y devuelve su índice o -1 si no está
def encontrar(arr, valor):
    # Recorremos el array con índices
    for i in range(arr.size):
        if arr[i] == valor:
            return i  # Devolvemos el índice si lo encontramos
    return -1  # Si no se encuentra, devolvemos -1

# Ejemplo de uso
datos = np.array([10, 20, 30, 40, 50])
print("Array:", datos)
print("Índice del valor 30:", encontrar(datos, 30))
print("Índice del valor 99:", encontrar(datos, 99))
```

4. Crear una función “desviación estándar” que reciba por parámetro un array y
devuelva la desviación estándar. Calcular paso a paso el resultado utilizando un for
para recorrer el array. También se debe usar la función “media” realizada
anteriormente. Para calcular la raíz cuadrada podemos usar la función de
numpy.sqrt(n). Operador potencia es **

```python
# Función para calcular la media aritmética
def media(arr):
    suma = 0
    for valor in arr:
        suma += valor
    return suma / arr.size

# Función para calcular la desviación estándar
def desviacion_estandar(arr):
    m = media(arr)  # Calculamos la media
    suma_cuadrados = 0
    # Sumamos los cuadrados de las diferencias respecto a la media
    for valor in arr:
        suma_cuadrados += (valor - m) ** 2
    varianza = suma_cuadrados / arr.size  # Calculamos la varianza
    return np.sqrt(varianza)  # Raíz cuadrada de la varianza

# Ejemplo de uso
datos = np.array([2, 4, 4, 4, 5, 5, 7, 9])
print("Array:", datos)
print("Desviación estándar:", desviacion_estandar(datos))
```

5. Crea una función reciba por parámetro un array y un valor y devuelva cuantas veces
aparece ese valor en el array.

```python
# Función que cuenta cuántas veces aparece un valor en el array usando NumPy
def contar_valor(arr, valor):
    # Creamos un array booleano donde cada elemento indica si es igual al valor buscado
    iguales = arr == valor
    # Sumamos los True (que equivalen a 1) para contar cuántas veces aparece
    return np.sum(iguales)

# Ejemplo de uso
datos = np.array([1, 2, 3, 2, 4, 2, 5])
print("Array:", datos)
print("Veces que aparece el 2:", contar_valor(datos, 2))
```

6. Crea una función que reciba por parámetro un array y devuelva cuantas veces
aparece el valor máximo en el array.

```python
# Función que cuenta cuántas veces aparece el valor máximo en el array
def contar_maximo(arr):
    max_val = np.max(arr)           # Obtenemos el valor máximo
    iguales = arr == max_val        # Array booleano: True donde el valor coincide
    return np.sum(iguales)          # Sumamos los True para contar coincidencias

# Ejemplo de uso
datos = np.array([3, 7, 7, 2, 7, 1])
print("Array:", datos)
print("Veces que aparece el máximo:", contar_maximo(datos))
```

7. Crea una función que reciba por parámetro un array y devuelva la moda.

```python
# Función que devuelve la moda de un array
# Podemos implementarla con unique + argmax
def moda(arr):
    # Obtenemos los valores únicos y sus frecuencias
    valores, frecuencias = np.unique(arr, return_counts=True)
    # Buscamos el índice del valor con mayor frecuencia
    indice_moda = np.argmax(frecuencias)
    # Devolvemos el valor correspondiente
    return valores[indice_moda]

# Ejemplo de uso
datos = np.array([1, 2, 3, 3, 3, 3, 4])
print("Array:", datos)
print("Moda:", moda(datos))
```

8. Crea una función que reciba por parámetro un array y devuelva otro array que
contenga los índices de aquellos valores que son mayores a la media.

```python
# Función que devuelve los índices de los valores mayores a la media
def indices_mayores_media(arr):
    m = np.mean(arr)            # Calculamos la media
    mayores = arr > m           # Array booleano: True donde el valor es mayor a la media
    return np.where(mayores)[0] # Extraemos los índices donde la condición se cumple

# Ejemplo de uso
datos = np.array([1, 4, 6, 2, 7])
print("Array:", datos)
print("Media:", np.mean(datos))
print("Índices de valores mayores a la media:", indices_mayores_media(datos))
```
