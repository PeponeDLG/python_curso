1. Crear una función “inicializar_array” que reciba por parámetro un array y devuelva el
array inicializado con valores introducidos por teclado.

```python
import numpy as np

# Inicializa un array de ceros del tamaño especificado por teclado
num_elementos = int(input("Introduzca el número de elementos del array: "))
array_ceros = np.zeros(num_elementos, int)

# Devuelve un array con los valores introducidos por teclado
def inicializar_array(array):
    for i in range(array.size):
        array[i] = input(f"Introduzca el elemento {i}: ")

    return array

# Prueba la función
array = inicializar_array(array_ceros)
print(f"Array: {array}")
print(f"Propiedades:")
print(f"Dimensiones: {array.ndim}")
print(f"Tamaño: {array.size}")
print(f"Shape: {array.shape}")
print(f"Tipo: {array.dtype}")
```

2. Crear las funciones “media”, “máximo”, “mínimo” que reciba un array de valores
numéricos y devuelva la media aritmética, el valor máximo y el valor mínimo
respectivamente. Realizar el cálculo manualmente recorriendo el array con un for.

```python
# Devuelve la media de los valores de un array
def media(array):
    sum = 0
    for i in range(array.size):
        sum += array[i]

    return sum / array.size

# Devuelve el valor máximo de un array
def maximo(array):
    max = 0
    for i in range(array.size):
        if array[i] > max:
            max = array[i]

    return max

# Devuelve el valor mínimo de un array
def minimo(array):
    min = array[0]
    for i in range(array.size):
        if array[i] < min:
            min = array[i]

    return min

# Muestra el array
print(f"Array: {array}")

# Prueba las funciones
print(f"Media: {media(array)}")
print(f"Valor máximo: {maximo(array)}")
print(f"Valor mínimo: {minimo(array)}")
```

3. Crear la función “encontrar” (usando for) que reciba dos parámetros: un array
numérico y un valor. La función debe buscar en el array si el “valor” se encuentra en
él. La función debe devolver:
- Si valor se encuentra en el array, devolver el índice (primera aparición)
- Si el valor no se encuentra en el array, devolver -1.

```python
# Busca un valor dado en un array. Si lo encuentra devuelve el índice, y si no, -1
def encontrar(array, valor):
    for i in range(array.size):
        if array[i] == valor:
            return i

    return -1

# Prueba la función
print(f"Array: {array}")
print(f"Índice del elemento con valor '3': {encontrar(array, 3)}")
```

4. Crear una función “desviación estándar” que reciba por parámetro un array y
devuelva la desviación estándar. Calcular paso a paso el resultado utilizando un for
para recorrer el array. También se debe usar la función “media” realizada
anteriormente. Para calcular la raíz cuadrada podemos usar la función de
numpy.sqrt(n). Operador potencia es **

```python
# Devuelve la desviación estándar
# 1. Calcula la suma del cuadrado de la desviación de cada dato respecto a la media
# 2. Divide el resultado entre el número de datos obteniendo la varianza y devuelve su raíz cuadrada
def desviacion_estandar(array):
    sum = 0
    for i in range(array.size):
        sum += (array[i] - media(array)) ** 2

    return np.sqrt(sum / array.size)

# Prueba la función
print(f"Array: {array}")
print(f"Desviación estándar: {desviacion_estandar(array)}")
```

8. Crea una función que reciba por parámetro un array y devuelva otro array que
contenga los índices de aquellos valores que son mayores a la media.

```python
# Devuelve el número de ocurrencias de un valor dado en un array
def encontrar(array, valor):
    return len(np.where(array == valor)) # where() devuelve una tupla con los índices de los elementos coincidentes

# Devuelve el número de ocurrencias del valor máximo de un array
def num_veces_valor_max(array):
    return encontrar(array, np.max(array))

# Devuelve la moda o modas de los valores de un array. Si no hay, devuelve null
# Nota: No hay una función para la moda en NumPy pero sí en SciPy. Para no usar SciPy antes de verla en clase,
# hago el cálculo por partes:
# 1. Crea un diccionario con key=valores del array (omitiendo los repetidos) y value=0.
# 2. Añade como valores del diccionario las ocurrencias de cada valor del array.
# 3. Detecta el valor o valores máximos del diccionario. Si hay más de uno, los añade a una lista.
# 4. Devuelve el valor (unimodal) o lista de valores (multimodal), o None si no hay moda.
def moda(array):
    diccionario = dict.fromkeys(array, 0) # 1
    for clave, valor in diccionario.items(): # 2
        for elem in array:
            if clave == elem:
                diccionario[clave] += 1

    lista_maximos = []
    max_repeticiones = max(diccionario.values()) # Máximo de veces que se repite un valor de entre los valores del array
    for clave, valor in diccionario.items(): # 3
        if valor == max_repeticiones:
            lista_maximos.append(clave)
    
    if len(lista_maximos) == array.size: # 4 (No hay moda)
        return None
    elif len(lista_maximos) > 1: # 4 (Multimodal)
        modas = []
        for elem in lista_maximos:
            modas.append(int(elem))
        return modas
    else: # 4 (Unimodal)
        return int(lista_maximos[0])

# Recibe un array y devuelve otro array con los índices de los valores que superan la media
def superiores_media(array):
    indices = []
    media = np.mean(array)
    for i in range(array.size):
        if array[i] > media:
            indices.append(i)

    return np.array(indices)

# Prueba las funciones
print(f"Array: {array}")
print(f"Número de ocurrencias del entero 3: {encontrar(array, 3)}")
print(f"Número de ocurrencias del valor máximo: {num_veces_valor_max(array)}")
print(f"Moda o modas: {moda(array)}")
print(f"Array con los índices de los elementos superiores a la media: {superiores_media(array)}")
```
