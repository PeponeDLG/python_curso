**Cargamos librerías y datos**

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("archivo_completo.csv")
```

**Carga y exploración estructural**

```python
print("\nDimensiones del dataset")
print("-----------------------\n")
fila_columna = data.shape
print(f"Filas -> {fila_columna[0]}")
print(f"Columnas -> {fila_columna[1]}")

print("\nInicio")
display(data.head(10))
print("Fin")
display(data.tail(10))
```

Vemos que la primera columna es "Unnamed: 0" y contiene las fechas/horas.  
Vamos a renombrarla y limpiar los nombres de las columnas para trabajar más cómodamente.

```python
# Renombramos la columna de fecha
data.rename(columns={"Unnamed: 0":"Fecha_Hora"}, inplace=True)

# Convertimos Fecha_Hora a datetime
data["Fecha_Hora"] = pd.to_datetime(data["Fecha_Hora"])

# Limpiamos los nombres de las columnas: quitamos espacios, corchetes y caracteres especiales
data.columns = [col.strip().replace(" ", "_").replace("[","").replace("]","").replace("__","_") for col in data.columns]

print("\nColumnas tras la limpieza:")
print(data.columns.tolist())
print("\nTipos de datos:")
print(data.dtypes)
```

**Tratamiento de duplicados**

```python
print("Registros totales: ", len(data))
print("Registros duplicados: ", data.duplicated().sum())
data = data.drop_duplicates()
print("Registros tras eliminar duplicados: ", len(data))
```

**Tratamiento de nulos**

```python
nulos = data.isna().sum()
nulos = nulos[nulos > 0]

if nulos.__len__() <= 0:
    print("\n -- NO HAY NULOS --\n")
else:
    print("\nColumnas con nulos:")
    display(nulos)
```

**Optimización de tipos de datos**  
Convertimos las columnas numéricas a float32 para optimizar memoria.

```python
# Identificamos columnas numéricas (excluyendo Fecha_Hora)
cols_numericas = data.select_dtypes(include=['float64', 'int64']).columns
for col in cols_numericas:
    data[col] = data[col].astype('float32')

print("Memoria usada:", data.memory_usage(deep=True).sum() / 1024**2, "MB")
print("\nTipos actualizados:")
print(data.dtypes)
```

**Fase 2: Perfilado y Detección de Anomalías**

Resumen estadístico de las variables numéricas

```python
display(data.describe().round(2))
```

**Sensores de Humedad del Suelo**  
Vemos que los valores de humedad oscilan entre ~0% y ~100%, con medias alrededor del 20-30%.  
Algunos sensores tienen mínimos de 0.0, lo que podría indicar lecturas erróneas o periodos de sequía extrema.  
  
**Sensores de Temperatura del Suelo**  
Las temperaturas medias rondan los 23-24°C, con rangos que van desde ~10°C hasta ~40°C.  
  
**Batería y Panel Solar**  
La batería tiene un mínimo de 0 mV, lo que indica que en algún momento se agotó por completo.  
El panel solar también tiene valores a 0, probablemente durante la noche.

**Análisis de Distribución -> Humedad del Suelo (Sensor 1)**

```python
col_humedad = "sensor_de_humedad_del_suelo_1_%"
print("Media: ", data[col_humedad].mean())
print("Mediana: ", data[col_humedad].median())
print("Diferencia media-mediana: ", round(float(abs(data[col_humedad].mean()-data[col_humedad].median())), 10))
print("Fisher (asimetría): ", data[col_humedad].skew())
print("Curtosis: ", data[col_humedad].kurt())
```

La asimetría positiva indica que la mayoría de las lecturas de humedad se concentran por debajo de la media.  
La curtosis nos indica la presencia de valores extremos (outliers).  
  
**Detectemos los outliers con un boxplot.**

```python
fig1, ax1 = plt.subplots(figsize=(12, 3))
ax1.boxplot(data[col_humedad],
            vert=False,
            patch_artist=True,
            boxprops=dict(color='red'),
            medianprops=dict(color='red'),
            whiskerprops=dict(color='blue'),
            capprops=dict(color='blue'),
            flierprops=dict(markerfacecolor='orange', markersize=8))
plt.title("Diagrama de bigotes - Humedad del Suelo Sensor 1")
plt.xlabel("Humedad [%]")
plt.show()
```

Se observan varios outliers por encima del bigote superior.  
Estos valores extremos podrían deberse a lecturas después de lluvias intensas o a errores de medición.

**Análisis de Distribución -> Temperatura del Suelo (Sensor 1)**

```python
col_temp = "sensor_de_temperatura_del_suelo_1_°C"
print("Media: ", data[col_temp].mean())
print("Mediana: ", data[col_temp].median())
print("Diferencia media-mediana: ", round(float(abs(data[col_temp].mean()-data[col_temp].median())), 10))
print("Fisher (asimetría): ", data[col_temp].skew())
print("Curtosis: ", data[col_temp].kurt())
```

```python
fig2, ax2 = plt.subplots(figsize=(12, 3))
ax2.boxplot(data[col_temp],
            vert=False,
            patch_artist=True,
            boxprops=dict(color='red'),
            medianprops=dict(color='red'),
            whiskerprops=dict(color='blue'),
            capprops=dict(color='blue'),
            flierprops=dict(markerfacecolor='orange', markersize=8))
plt.title("Diagrama de bigotes - Temperatura del Suelo Sensor 1")
plt.xlabel("Temperatura [°C]")
plt.show()
```

La temperatura tiene una distribución más simétrica que la humedad, con pocos outliers.  
Esto es esperable ya que la temperatura del suelo tiende a ser más estable.

**Análisis de Batería y Panel Solar**

```python
print("\nEstadísticas de Batería [mV]")
display(data["Batería_mV"].describe().round(2))

print("\nEstadísticas de Panel Solar [mV]")
display(data["Panel_solar_mV"].describe().round(2))
```

**Fase 3: Identificar Cómplices (Análisis de Relaciones)**

**Matriz de Correlación entre sensores de humedad**

```python
# Seleccionamos solo las columnas de humedad
cols_humedad = [col for col in data.columns if "humedad" in col]
corr_humedad = data[cols_humedad].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_humedad, annot=True, cmap="coolwarm", linewidths=0.5, fmt=".2f")
plt.title("Mapa de calor - Correlación entre sensores de humedad")
plt.show()
```

Observamos una alta correlación positiva entre todos los sensores de humedad, lo que indica que todos miden patrones similares.  
Esto tiene sentido ya que están en la misma zona geográfica y responden a los mismos eventos de lluvia/riego.

**Matriz de Correlación entre sensores de temperatura**

```python
cols_temp = [col for col in data.columns if "temperatura" in col]
corr_temp = data[cols_temp].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_temp, annot=True, cmap="coolwarm", linewidths=0.5, fmt=".2f")
plt.title("Mapa de calor - Correlación entre sensores de temperatura")
plt.show()
```

Igual que con la humedad, los sensores de temperatura están altamente correlacionados entre sí.  
Esto sugiere que podríamos reducir dimensionalidad usando, por ejemplo, el promedio de todos los sensores.

**Relación entre Humedad y Temperatura (promedio de todos los sensores)**

```python
# Creamos columnas promedio
data["Humedad_Promedio"] = data[cols_humedad].mean(axis=1)
data["Temperatura_Promedio"] = data[cols_temp].mean(axis=1)

plt.figure(figsize=(10, 6))
plt.scatter(data["Temperatura_Promedio"], data["Humedad_Promedio"],
            alpha=0.3, c='blue', s=10)
plt.xlabel("Temperatura del Suelo Promedio [°C]")
plt.ylabel("Humedad del Suelo Promedio [%]")
plt.title("Relación entre Humedad y Temperatura del Suelo")
plt.grid(True, alpha=0.3)
plt.show()
```

Se observa una ligera correlación negativa: a mayor temperatura, menor humedad.  
Esto es coherente con el comportamiento esperado: cuando hace más calor, el suelo se seca más rápido.

**Evolución temporal de las variables**

```python
fig, axes = plt.subplots(3, 1, figsize=(16, 10))

# Humedad promedio en el tiempo
axes[0].plot(data["Fecha_Hora"], data["Humedad_Promedio"], color='green', alpha=0.7, linewidth=0.5)
axes[0].set_title("Evolución de la Humedad del Suelo (Promedio)")
axes[0].set_ylabel("Humedad [%]")
axes[0].grid(True, alpha=0.3)

# Temperatura promedio en el tiempo
axes[1].plot(data["Fecha_Hora"], data["Temperatura_Promedio"], color='red', alpha=0.7, linewidth=0.5)
axes[1].set_title("Evolución de la Temperatura del Suelo (Promedio)")
axes[1].set_ylabel("Temperatura [°C]")
axes[1].grid(True, alpha=0.3)

# Batería en el tiempo
axes[2].plot(data["Fecha_Hora"], data["Batería_mV"], color='purple', alpha=0.7, linewidth=0.5)
axes[2].set_title("Evolución de la Batería")
axes[2].set_ylabel("Batería [mV]")
axes[2].set_xlabel("Fecha / Hora")
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

**Análisis de la Batería vs Panel Solar**  
Veamos cómo se relaciona la carga de la batería con la generación del panel solar.

```python
plt.figure(figsize=(10, 6))
plt.scatter(data["Panel_solar_mV"], data["Batería_mV"],
            alpha=0.3, c='orange', s=10)
plt.xlabel("Panel Solar [mV]")
plt.ylabel("Batería [mV]")
plt.title("Relación entre Panel Solar y Batería")
plt.grid(True, alpha=0.3)
plt.show()
```

Se observa una relación directa: cuando el panel solar genera más energía, la batería está más cargada.  
Los valores de panel solar a 0 mV corresponden a periodos nocturnos donde la batería se va descargando progresivamente.

**Matriz de correlación completa (variables principales)**

```python
cols_principales = ["Humedad_Promedio", "Temperatura_Promedio", "Batería_mV", "Panel_solar_mV"]
corr_matrix = data[cols_principales].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", linewidths=0.5, fmt=".3f")
plt.title("Mapa de calor - Correlación entre variables principales")
plt.show()
```
