```python
import matplotlib.pyplot as plt
import numpy as np
arr1 = np.random.standard_normal(50)
```

```python
#Mostramos el array generado
print("Array aleatorio generado:")
print(arr1)
```

```python
#plot de línea
#La siguiente línea crea una figura o "lienzo" sobre el quese dibujarán los gráficos
#Se fila un tamaño de 10 pulgadas deancho y 10 de alto
plt.figure(figsize=(10,4))
plt.plot(arr1,color='darkorange', linewidth=2, linestyle='--', marker='o')
plt.title("Gráfico de línea de arr1")
#Etiquetas de los ejes
plt.ylabel("Índice")
plt.xlabel("Valor")
#Podemos activar la cuadrícula para facilitar la lectura de los valores.
plt.grid(True)
# Ajusta automáticamente los espacios y márgenes de la figura.
#Evita que las etiquetas de los ejes, títulos o leyendas se solapen entre sí o queden cortadas.
#Es especialmente útil cuando tienes varios subplotsAjusta automáticamente los espacios y márgenes de la figura.
plt.tight_layout()
plt.show()
```

2.Crea un histograma de arr1

```python
#plot de línea
#La siguiente línea crea una figura o "lienzo" sobre el quese dibujarán los gráficos
#Se fila un tamaño de 10 pulgadas deancho y 10 de alto
plt.figure(figsize=(8,4))
plt.hist(arr1,bins=10,color='darkorange', edgecolor='black')
plt.title("Histograma de arr1")
#Etiquetas de los ejes
plt.ylabel("Valor")
plt.xlabel("Frecuencia")
#Podemos activar la cuadrícula para facilitar la lectura de los valores.
plt.grid(True)
# Ajusta automáticamente los espacios y márgenes de la figura.
#Evita que las etiquetas de los ejes, títulos o leyendas se solapen entre sí o queden cortadas.
#Es especialmente útil cuando tienes varios subplotsAjusta automáticamente los espacios y márgenes de la figura.
plt.tight_layout()
plt.show()
```

Calcula la media, la mediana, la desviación, percentil 25 y 75, compara con el histograma

```python
media=np.mean(arr1)
mediana=np.median(arr1)
desviacion=np.std(arr1)
p25=np.percentile(arr1,25)
p75=np.percentile(arr1,75)
print("La media es: ", media)
print("La mediana es: ", mediana)
print("La desviación estándar es: ", desviacion)
print("El percentil 25 es: ", p25)
print("El percentil 75 es: ", p75)
```

Comparación con el histograma

Explicación:Tanto la media como la mediana están próximas a 0, lo que indica cierta concentración entorno a dicho valor (también se refleja en el histograma)
Con respecto a la desviación estándar, se deduce que la mayoría de los datos se encuentran dentro de un rango aproximado de 
(-0.13-1,125,-0.13+1,125). Observando el histograma llegamos a los mismos resultados porque hay una mayor densidad de barras en ese intervalo.
En cuanto a los percentiles: El 25% de la muestra está por debajo del valor -0,66 y el 75%
