```python
import numpy as np
import matplotlib.pyplot as mp

# =====================================================================
# CARGA DEL ARCHIVO (PASO NECESARIO PARA TODAS LAS PREGUNTAS)
# =====================================================================

# Utilizamos np.genfromtxt() para cargar el archivo CSV.
# Se utiliza ',' como delimitador y se asume que los datos son homogéneos (mayormente numéricos),
# permitiendo que NumPy infiera el tipo de dato o use float por defecto [12, 13].

try:
    datos = np.genfromtxt("datos_estudiantes.csv", delimiter=',')
    print("--- Archivo cargado exitosamente ---")
except FileNotFoundError:
    print("Error: Asegúrate de que 'datos_estudiantes.csv' existe en el directorio.")
    # Si la carga falla, el script no puede continuar
    exit()

# Definición de las columnas para mayor claridad
ID = datos[:, 0]
NOTAS_MATES = datos[:, 1]
NOTAS_FISICA = datos[:, 2]
EDADES = datos[:, 3]
```

1. Inspección del Array: Muestra el número total de elementos que contiene el array, el número de
dimensiones, y la forma o tamaño por dimensión

```python
print("Muestra: \n", datos[:5])
print("Tipo: ", datos.dtype)
print("Nº de elementos:",datos.size)
print("Dimensiones:",datos.ndim)
print("Tamaño:",datos.shape)
```

2. Filtrado por Condición: Utilizando indexación booleana, filtra el array para obtener un nuevo array que
contenga solamente las notas de Matemáticas (Columna 1) de aquellos estudiantes cuya Edad (Columna 3)
sea menor o igual a 20.

```python
menores_20 = NOTAS_MATES[EDADES <= 20]
print("Notas de los alumnos de 20 o menos años:",menores_20,"-> tiene ",menores_20.size,"elementos")
```

1. Cálculo Agregado: Calcula la media aritmética y la desviación estándar de todas las Notas de Física.

```python
media_fisica = np.mean(NOTAS_FISICA)
desv_estandar = np.std(NOTAS_FISICA)
print("Media de las notas de física:",media_fisica)
print("Desv de las notas de física:",desv_estandar)
```

2. Percentil: Determina el valor de la nota que se encuentra en el percentil 75 del conjunto de datos de Notas

```python
print("Percentil 75:",np.percentile(NOTAS_FISICA,75))
```

Pregunta 5: Comparación de Arrays y Mínimos/Máximos
Considere las columnas de Notas de Matemáticas (Columna 1) y Notas de Física (Columna 2):
1. Ufunc Binaria: Genera un nuevo array llamado nota_minima_por_alumno que contenga la nota mínima
obtenida por cada estudiante entre Matemáticas y Física, utilizando una ufunc binaria de NumPy.
2. Conjunto Único de Suspendidos: Determina cuántos valores de notas únicos (sin repetir) existen en la columna
de Notas de Física que sean menores a 5.

```python
nota_minima_por_alumno = np.where(NOTAS_MATES < NOTAS_FISICA,NOTAS_MATES,NOTAS_FISICA)
nota_minima_por_alumno = np.minimum(NOTAS_MATES,NOTAS_FISICA)

print("mínimos:",nota_minima_por_alumno)
```
