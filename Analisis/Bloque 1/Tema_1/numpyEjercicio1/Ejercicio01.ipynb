*1.* Crea un array a partir de una lista de números con decimales (los que tu consideres, 4 o
5 números). Imprime sus valores uno a uno mediante un bucle while y los atributos del array
(dimensiones, tamaño, tipo de dato, shape).

```python
import numpy as np
array1 = np.array([0.5,0.8,3.6,4.4,5.0])
i = 0

while i < array1.size:
    print(array1[i])
    i += 1

print("Nº dimensiones: ", array1.ndim)
print("Tamaño: ", array1.size)
print("Tipo de dato: ",array1.dtype)
print("Tipo de array", array1.shape)
```

2. Crear una función “inicializar_array” que reciba por parámetro un array de 1 dimensión de
tipos enteros y devuelva el array inicializado/modificado con valores introducidos por teclado
(solamente enteros). Utilizar un while para recorrer el array.

```python
def inicializar_array(s):
    arrayAux = np.zeros(s,int)
    i = 0
    #print(arrayAux.dtype)
    
    while i < arrayAux.size:
        arrayAux[i] = int(input("Introduzca número de la posición "))
        i += 1
    return arrayAux
    
size = int(input("Introduzca tamaño del array: "))

array2 = inicializar_array(size)
i = 0

while i < array2.size:
    print(array2[i])
    i += 1

print("Nº dimensiones: ", array2.ndim)
print("Tamaño: ", array2.size)
print("Tipo de dato: ",array2.dtype)
print("Tipo de array", array2.shape)



        
```

3. Crear las funciones “media”, “máximo”, “mínimo” que reciba un array de valores
numéricos y devuelva la media aritmética, el valor máximo y el valor mínimo
respectivamente. Realizar el cálculo manualmente recorriendo el array con un while.

```python
def media(arr):
    # return np.mean(arr)
    i = a = 0
    
    while i < arr.size:
        a += arr[i]
        i += 1
    return a/arr.size

def maximo(arr):
    return np.max(arr)

def mediana(arr):
    return np.median(arr)

array3 = np.array([5,6,4.5,4,87,56,5,456,4,6.6,7.98])

print("Media: ",media(array3))
print("Mediana: ",mediana(array3))
print("Máximo: ",maximo(array3))
```

4. Crear la función “encontrar” (usando while) que reciba dos parámetros: un array numérico
y un valor. La función debe buscar en el array si el “valor” se encuentra en él. La función
debe devolver:
○ Si valor se encuentra en el array, devolver el índice (primera aparición)
○ Si el valor no se encuentra en el array, devolver -1

```python
def encontrar(arr, x):
    i = 0
    a = -1
    
    while i < arr.size:
        if x == arr[i]:
            a = i
            break
        i += 1
        
    return a

array4 = np.array([5,6,4.5,4,87,56,5,456,4,6.6,7.98])
x = 4
print("El valor ",x," está en la posición: ",encontrar(array4,x))
```

5. Crear una función “desviación estándar” que reciba por parámetro un array y devuelva la
desviación estándar. Calcular paso a paso el resultado utilizando un while para recorrer el
array. También se debe usar la función “media” del ejercicio 2. Para calcular la raíz
cuadrada podemos usar la función de numpy.sqrt(n). Operador potencia es **

```python
def media(arr):
    # return np.mean(arr)
    i = a = 0
    
    while i < arr.size:
        a += arr[i]
        i += 1
    return a/arr.size
    
def desvEst(arr, m):
    # return np.mean(arr)
    i = suma = 0
    
    while i < arr.size:
        suma += np.square(arr[i] - m)
        i += 1
    
    return np.sqrt(suma/arr.size)

array5 = np.array([5,6,4.5,4,87,56,5,456,4,6.6,7.98])

print("La desviación estandar es: ",desvEst(array5,media(array5)))
print("La desviación estandar np: ",np.std(array5))
```
