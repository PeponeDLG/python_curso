```python
import numpy as np

# =====================================================================
# CARGA DEL ARCHIVO (PASO NECESARIO PARA TODAS LAS PREGUNTAS)
# =====================================================================

# Utilizamos np.genfromtxt() para cargar el archivo CSV.
# Se utiliza ',' como delimitador y se asume que los datos son homogéneos (mayormente numéricos),
# permitiendo que NumPy infiera el tipo de dato o use float por defecto [12, 13].

try:
    datos = np.genfromtxt("datos_estudiantes.csv", delimiter=',')
    print("--- Archivo cargado exitosamente ---")
except FileNotFoundError:
    print("Error: Asegúrate de que 'datos_estudiantes.csv' existe en el directorio.")
    # Si la carga falla, el script no puede continuar
    exit()

# Definición de las columnas para mayor claridad
ID = datos[:, 0]
NOTAS_MATES = datos[:, 1]
NOTAS_FISICA = datos[:, 2]
EDADES = datos[:, 3]
```
