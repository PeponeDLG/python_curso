```python
import numpy as np
nombres = np.array(["Alberto", "Álvaro", "Antonio", "Guillermo", "José Manuel", "Juan Diego", "Juan Ignacio","Juan José", "María", "Raúl", "Victor"])
print(nombres)
notas = np.random.randint(1, 11, nombres.size)
print(notas)
```

```python
print("1. Nota media de los alumnos: ",notas.mean())
```

```python
print("2. Desviación estandar: ",notas.std())
```

```python
print("3. Número de aprobados: ",sum(notas >= 5))
```

```python
print("4. Nombres de aprobados: ")
alumnos = nombres[notas >= 5]

for i in alumnos:
    print("- ",i)
```

```python
print("5. Nombres con nota superior a la media (",notas.mean(),"): ")
alumnos = nombres[notas > notas.mean()]

for i in alumnos:
    print("- ",i)
```

```python
print("6. Mostrar nombres del revés mediante una slice: ")
reves = nombres[::-1].copy()

for i in reves:
    print("- ",i)
```

```python
print("7. Ordenar, haciendo una copia de datos, las notas de menor a mayor")
orden = np.sort(notas.copy())

for i in orden:
    print("- ",i)
```

```python
print("8. Asignar un 5 a los que tienen un 4")
notas[notas == 4] = 5

for i in notas:
    print("- ",i)
```

```python
print("9. Sumar 1 a todas las notas: ")
# sup = np.add(notas,np.ones(notas.size))
sup = notas + 1

for i in sup:
    print("- ",i)
```

```python
print("10. Sumar 1 a todas las notas mayores a 4: ")

# sup = notas[notas > 4]
# sup = np.add(sup,np.ones(sup.size))

sup = notas[notas > 4].copy() + 1


for i in sup:
    print("- ",i)
```
