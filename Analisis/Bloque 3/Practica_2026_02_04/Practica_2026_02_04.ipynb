**Carga de CSVs de datos meteorológicos**
El dataset que he elegido es muy grande y está dividido en múltiples .csv, por lo que tengo que hacer un algoritmo que recorra todos los ficheros dentro de la carpeta  
donde tiene

```python
import os
import pandas as pd
import numpy as np
import matplotlib as mp
import matplotlib.pyplot as plt


path = os.path.join(os.getcwd(),"datos/Datos_meteorológicos_por_hora_por_municipio_España/Montly-wheather-data-2015-2020")
lFicheros = os.listdir(path) # listamos todos los csv que hay en la carpeta

print(path)
lFicheros.sort() # ordenamos la lista de ficheros csv a cargar para cargarlos de antiguos a actuales

df1 = pd.Series()

for f in lFicheros:
    print(f)
    df1 = pd.read_csv(os.path.join(path,f),delimiter=";")
    break # descomentar cuando acabemos las pruebas

print(f"Número de registros en s1: {df1["id"].count()}")

for c in df1.axes[1].array:
    print(c, end=" - ")
```

**Leyenda de columnas**  
__fecha__ -> fecha de la recogida de los datos  
**G(i)** -> Irradiancia Global en el plano inclinado. Radiación solar por unidad de área (W/m²) que incide sobre una superficie  
**H_sun** -> Horas de Sol durante la cual la radiación solar directa supera un umbral de intensidad  
**T2m** -> Es la Temperatura del Aire a 2 metros. Medida a 2 metros sobre el suelo para evitar la influencia directa de la radiación terrestre  
**WS10m** -> Velocidad del Viento a 10 metros. Es la altura estándar para la medición de la velocidad del viento en meteorología.
**id** -> Id del municipio, que está especificado en el fichero fronteras_municipios.csv
