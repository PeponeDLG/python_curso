**2. Diagnóstico de Calidad y Gestión de Nulos (RA2.b, RA2.c)**  
  
• Detecta la presencia de valores ausentes (NaN) en todo el DataFrame mediante las funciones isna() o notna().  
• Tratamiento de nulos en "ciudad": Dado que los trabajos remotos pueden no tener ciudad, rellena los valores faltantes en esta columna  con la etiqueta "Sin especificar/Remoto" usando un valor escalar en fillna().  
• Tratamiento de nulos en "min_años_de_experiencia": Sustituye los valores nulos por la mediana de los años de experiencia para evitar  distorsiones en cálculos matemáticos posteriores.

```python
display(data.head(10))
print("¿Tiene nulos? ->", np.where(np.any(data.isna()), "Si", "No")) # Indicamos si tiene o no nulos con isna()
print("¿Tiene nulos? ->", np.where(np.all(data.notna()), "No", "Si")) # Indicamos si tiene o no nulos con isna()

print("\nColumnas con nulos:\n")
for col in data:
    print("Col.", col, " -> ", np.where(np.any(data[col].isna()),"Si","No"))

print("\n=====================================\n")
print("Sustitución de NaN por Remote\n")
data.city = data.city.fillna("Remote")
data.min_experience_years = data.min_experience_years.fillna(data.min_experience_years.mean())
display(data.head(10))

print("¿Tiene nulos? ->", np.where(np.any(data.isna()), "Si", "No")) # Indicamos si tiene o no nulos con isna()
print("¿Tiene nulos? ->", np.where(np.all(data.notna()), "No", "Si")) # Indicamos si tiene o no nulos con isna()

print("\nColumnas con nulos:\n")
for col in data:
    print("Col.", col, " -> ", np.where(np.any(data[col].isna()),"Si","No"))
```

**3. Limpieza de Redundancias e Integridad (RA2.c)**  
• Verifica si existen ofertas de empleo duplicadas basándote en la columna id_del_trabajo

• Identifica columnas que no aporten valor al análisis estadístico (como id_del_trabajo si ya no

**4. Transformación y Conversión de Datos (RA2.e, RA2.f)**  
• Normalización de Fechas: Convierte la columna año_publicado al tipo de dato datetime de Pandas usando pd.to_datetime() para facilitar  futuros análisis temporales.  
• Cálculo de Variables Cuantitativas: Crea una nueva columna llamada salario_promedio_usd que sea el resultado de la media aritmética  entre salario_mínimo_usd y salario_máximo_usd.  
• Conversión de Categóricos: Identifica las columnas de tipo object (como tamaño de la empresa) y conviértelas al tipo de dato category  para optimizar el rendimiento y la memoria.

```python
data["posted_year"] = pd.to_datetime(data["posted_year"],format="%Y")

# data["salario_promedio_usd"] = pd.Series((data.salary_max_usd + data.salary_min_usd)/2) -> funciona pero no me gusta
data["salario_promedio_usd"] = data[["salary_min_usd","salary_max_usd"]].mean(axis=1) # axis=1 significa en eje horizontal - axis=0 es el eje vertical

#display(data)

# Primero hacemos un estudio de las columnas tipo object para ver si tiene sentido transformarlos a tipo category
for l in data.columns:
    aux = pd.Series
    if data[l].dtype == "object":
        aux = data[l].drop_duplicates()
        print(l," - ",data[l].dtype, " -> ", aux.__len__())
    # else:
    #     print(l," - ",data[l].dtype)
    
# Como no hay ningún campo object con muchos valores únicos tiene sentido transformarlos todos a category

for l in data.columns:
    if data[l].dtype == "object":
        data[l] = data[l].astype("category")

print("\n\n")
for l in data.columns:
    print(l," - ",data[l].dtype)
```

**5. Agrupación por Intervalos (Binning) (RA2.d)**  
• Utiliza la función pd.cut() para crear una nueva columna llamada rango_salarial. Divide los salarios en tres categorías: "Entrada", "Competitivo" y "Top", definiendo  manualmente los límites de los intervalos basándote en la columna salario_máximo_usd.

```python
aux = data.salary_max_usd.drop_duplicates() # serie para estudiar los valores de la columna de salarios máximos
max = aux.max()
min = aux.min()
med = int(aux.mean().__round__())
min_med = int((min + med)/2).__round__() # Para el rango "competitivo" hacemos la media entre salrio mínimo y la media de maximo y mínimo
max_med = int((max + med)/2).__round__() # y la media entre salrio máximo y la media de maximo y mínimo

print("Min: ",min)
print("Min med:",min_med)
print("Med: ",med)
print("Max med:",max_med)
print("Max: ",max)

print("\n","="*10,"\n")

data["rango_salarial"] = pd.cut(data.salary_max_usd,[min, min_med, max_med, max],labels=["Entrada", "Competitivo", "Top"])

display(data)
```
