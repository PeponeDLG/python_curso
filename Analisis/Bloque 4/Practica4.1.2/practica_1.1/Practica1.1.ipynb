```python
import os
import numpy as np
import matplotlib as mp
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv("./dist_renta_hogares.csv",delimiter=";")
```

```python
display(data)
```

**Vamos a ver las columnas que tienen nulos**

```python
print("¿Tiene nulos?")
cols_null = []

for i in data.columns:
    nulo = np.any(data[i].isna())
    
    print(f"{i} -> {np.where(nulo, 'Si', 'No')}")
    
    [cols_null.append(i) if nulo else None] # guardamos los nombres de las columnas para estudiarlas

print("\n")
print("\nConteo de valores únicos\n")

for i in data.columns:
    if i in cols_null:
        unic = data[i].drop_duplicates()
        print(f"{i} - {unic.count()} - {data[i].dtype}")
        cont = 0
        # for j in unic:
        #     print(j)            
        #     if cont < 100: cont += 1
        #     else: break


print("\n- Viendo los resultados podemos deducir que son columnas no categorizables -")
print("- \"Total\" es la única columna que podemos hacer la media para rellenar los nulos *Además hay que limpiar el carácter \".\" que impide que se haga la media-")

# Quitamos los nulos de Distritos y Secciones

data.Distritos = data.Distritos.fillna("No")
data.Secciones = data.Secciones.fillna("No")

# Quitamos los nulos de Total con la media de las celdas con valor numérico y entendemos que los . son el valor 0

mascara_punto = data.Total == "."
index_puntos = data.Total[mascara_punto]
mascara_nulo = data.Total == None
index_nulos = data.Total[mascara_nulo]

print("- Total puntos en [Col. Total] -> ", index_puntos.count())
print("- Total nulos en [Col. Total] -> ", index_nulos.count())

data.Total = data.Total.replace(",", ".", regex=True) # reemplazamos las , por .
data['Total'] = data['Total'].replace('^\\.$', 0, regex=True) # con regex cambiamos los . solitarios por valor 0 (esto si lo he mirado en la IA)
data.Total = data.Total.astype(float) # convertimos la columna en float
data.Total = data.Total.fillna(data.Total.mean().__round__(2)) # rellenamos los nulos con el valor medio

display(data)
```

**Veamos los tipos de las columnas y convertimos los object a category (excepto la columna Total)**

```python
for col in data.columns:
    print(f"{col} -> {data[col].dtype}")
    if data[col].dtype == "object" and col != data.Total.name:
        data[col] = data[col].astype("category")

print("\n Visualización del cambio \n")
for col in data.columns:
    print(f"{col} -> {data[col].dtype}")
```

**Agregamos una nueva columna con los códigos postales**

```python
data["Cod_Postal"] = str(data.Municipios).split()[1]
display(data)
```

**Normalizamos el periodo con datetime y ordenamos por Periodo y Codigo postal**

```python
data.Periodo = pd.to_datetime(data.Periodo, format="%Y")
data = data.sort_values(["Periodo","Cod_Postal"], ascending=[True ,False])
display(data)
```

**Categorizamos los valores de TOTAL por Bajo, Medio y Alto**

```python
aux = data.Total.drop_duplicates()
print(f"Entendemos que el porcentaje es de 0% a 100% entonces separamos cada 33,3%")

min = 0.0
min_med = 33.33
max_med = 66.66
max = 100.00

data["Porcentaje"] = pd.cut(data.Total, [min, min_med, max_med, max], labels=["Bajo", "Medio", "Alto"], right=False) # rigt=False para que coja el valor 0.0 como dentor del rango Bajo
# data["Porcentaje"] = data.Porcentaje.astype("category")

display(data)
```
