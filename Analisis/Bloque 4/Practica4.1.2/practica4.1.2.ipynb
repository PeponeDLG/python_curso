```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("./dist_renta_hogares_limpio.csv")
```

Comprobemos si los datos sirven para análisis

```python
print("Observamos esta columna que es clasificable en 3 categorías de umbral de probreza\n")
con_valor = data.Total > 0
con_valor = data[con_valor]
chorizo = data["Distribución de la renta por unidad de consumo"]
print(chorizo.value_counts())

# Cambiamos el nombre de una columna para trabajar mejor
data.rename(columns={"Distribución de la renta por unidad de consumo":"Distribucion_Renta"},inplace=True)


print("\nCreamos una nueva columna para clasificar con un código más corto los 3 umbrales de renta y hacemos recuento para ver que concuerda con la columna que categoriza\n")
# u5 = 5.000 - u7 = 7.500 - u10 = 10.000 
data["umbral_pobreza"] = np.where(data.Distribucion_Renta.str.contains("10.000"),"u10",np.where(data.Distribucion_Renta.str.contains("7.500"),"u7","u5") )
data["umbral_pobreza"] = data["umbral_pobreza"].astype("category")

print(data.umbral_pobreza.value_counts())

print("\n\n")

print("Como vemos, el periodo habla siempre del 1 de enero del año, por lo que simplificaremos el dato a solo el año\n")

data.Periodo = pd.to_datetime(data.Periodo)
data.Periodo = data.Periodo.dt.strftime("%Y")

print("\nCorregimos la columna Codigo_Posta, que vienen todas con el mísmo código \n")
Cod_Postal = np.where(data.Secciones != "No",data.Secciones.str[0:5],
                      np.where(data.Distritos != "No", data.Distritos.str[:5], data.Municipios.str[:5]))
data.Cod_Postal = Cod_Postal

print("\n\n")
display(data)
```

**Optimizamos los tipos de datos**

```python
data['Municipios'] = data['Municipios'].astype('category')
data['Distritos'] = data['Distritos'].astype('category')
data['Secciones'] = data['Secciones'].astype('category')
data['Sexo'] = data['Sexo'].astype('category')
data['Nacionalidad'] = data['Nacionalidad'].astype('category')
data['Periodo'] = data['Periodo'].astype('category')
data['Cod_Postal'] = data['Cod_Postal'].astype('category')

display(data.dtypes)
```
